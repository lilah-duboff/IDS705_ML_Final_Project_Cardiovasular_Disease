### Find Index Date

* Logic: The most recent date between earliest pair of sbp and scre record within 1 year timeframe from each other for each patient
* Purpose: Mark the start of follow up, make sure all lab records we will use for prediction is before this date and CVD event after this date to ensure fairness in risk prediction

```python
import pandas as pd
import numpy as np

# Fix SettingWithCopyWarning by creating explicit copies before modification
# or by using .loc for assignment operations

# Make explicit copies to avoid SettingWithCopyWarning
dat_sbp_copy = dat_sbp.copy()
dat_scre_copy = dat_scre.copy()

# Convert to datetime
dat_sbp_copy.loc[:, 'measurement_datetime'] = pd.to_datetime(dat_sbp_copy['measurement_datetime'])
dat_scre_copy.loc[:, 'measurement_datetime'] = pd.to_datetime(dat_scre_copy['measurement_datetime'])

# Select, drop duplicates, and rename in a single chain
dat_sbp = (dat_sbp_copy[['person_id', 'measurement_datetime', 'value_as_number']]
           .drop_duplicates()
           .rename(columns={'measurement_datetime': 'sbp_date', 'value_as_number': 'sbp_result'}))

dat_scre = (dat_scre_copy[['person_id', 'measurement_datetime', 'value_as_number']]
            .drop_duplicates()
            .rename(columns={'measurement_datetime': 'scre_date', 'value_as_number': 'scre_result'}))

# Create year timeframe - use loc for assignment to avoid warnings
dat_sbp.loc[:, 'year_range'] = pd.Timedelta(days=365)

# Find common patient IDs - use faster numpy-based intersection
sbp_ids = dat_sbp['person_id'].values
scre_ids = dat_scre['person_id'].values
pat_both = np.intersect1d(sbp_ids, scre_ids)

# Filter data using faster isin with numpy array
dat_sbp_filtered = dat_sbp[dat_sbp['person_id'].isin(pat_both)]
dat_scre_filtered = dat_scre[dat_scre['person_id'].isin(pat_both)]

# Create indices for faster merges
dat_sbp_filtered = dat_sbp_filtered.set_index('person_id').sort_index()
dat_scre_filtered = dat_scre_filtered.set_index('person_id').sort_index()

# Helper function for chunking
def process_in_chunks(ids, chunk_size=1000):
    num_chunks = (len(ids) + chunk_size - 1) // chunk_size
    for i in range(0, len(ids), chunk_size):
        chunk_ids = ids[i:i+chunk_size]
        print(f"Processing chunk {i//chunk_size + 1}/{num_chunks}")
        yield chunk_ids

# Process data in chunks
results = []
for chunk_ids in process_in_chunks(pat_both, chunk_size=1000):
    # Get data for current chunk
    df_sbp_tmp = dat_sbp_filtered.loc[chunk_ids].reset_index()
    df_scre_tmp = dat_scre_filtered.loc[chunk_ids].reset_index()
    
    # Merge data with inner join
    tmp = pd.merge(df_scre_tmp, df_sbp_tmp, on='person_id', how='inner')
    
    # Filter by date range
    mask = ((tmp['scre_date'] >= (tmp['sbp_date'] - tmp['year_range'])) & 
            (tmp['scre_date'] <= (tmp['sbp_date'] + tmp['year_range'])))
    tmp = tmp[mask]
    
    if tmp.empty:
        continue
    
    # Calculate time differences and minimize dates - use loc for new columns
    tmp.loc[:, 'col_diff'] = (tmp['scre_date'] - tmp['sbp_date']).abs()
    
    # Use numpy's minimum function for better performance
    tmp.loc[:, 'col_min'] = np.minimum(
        tmp['scre_date'].values.astype('datetime64[ns]'),
        tmp['sbp_date'].values.astype('datetime64[ns]')
    )
    
    # Sort and take first record for each patient
    tmp = tmp.sort_values(['person_id', 'col_min', 'col_diff'])
    tmp_sel = tmp.groupby('person_id', as_index=False).first()
    
    # Calculate index date using numpy for better performance
    tmp_sel.loc[:, 'INDEX_DATE'] = np.maximum(
        tmp_sel['scre_date'].values.astype('datetime64[ns]'),
        tmp_sel['sbp_date'].values.astype('datetime64[ns]')
    )
    
    # Select needed columns
    results.append(tmp_sel[['person_id', 'INDEX_DATE', 'scre_result', 'sbp_result']])

# Combine results if any exist
dat_index = pd.concat(results, ignore_index=True)
print(f"Number of patients with index date: {len(dat_index)}")
display(dat_index.head())
```

### Merge All Labs

```python
def merge_all_measurements(index_df, measurement_dfs, lookback_years=3):
    # copy
    result_df = index_df.copy()
    
    # convert to datetime
    result_df['INDEX_DATE'] = pd.to_datetime(result_df['INDEX_DATE']).dt.tz_localize(None)
    print("Preprocessing index dataframe...")
    
    # lookback window
    lookback_limits = result_df['INDEX_DATE'] - pd.DateOffset(years=lookback_years)
    
    # track progress
    total_measurements = len(measurement_dfs)
    measurement_count = 0
    
    # process each measurement type
    for measurement_type, measurement_df in measurement_dfs.items():
        measurement_count += 1
        print(f"\nProcessing {measurement_type} ({measurement_count}/{total_measurements})...")
        
        # convert to datetime
        meas_df = measurement_df.copy()
        meas_df['measurement_datetime'] = pd.to_datetime(meas_df['measurement_datetime']).dt.tz_localize(None)
        
        # init result columns
        result_df[f'{measurement_type}_value'] = np.nan
        result_df[f'{measurement_type}_date'] = pd.NaT
        result_df[f'{measurement_type}_days_diff'] = np.nan
        
        # store
        measurement_results = []
        
        # grouping
        print(f"  Grouping {measurement_type} measurements by person_id...")
        grouped_measurements = dict(list(meas_df.groupby('person_id')))
        
        # progress update
        total_rows = len(result_df)
        update_interval = max(1, total_rows // 20)
        
        print(f"  Processing {total_rows} people for {measurement_type} measurements:")
        
        for idx, (i, row) in enumerate(result_df.iterrows()):
            # show progress
            if idx % update_interval == 0 or idx == total_rows - 1:
                progress_pct = min(100, round((idx + 1) / total_rows * 100))
                print(f"  Progress: {progress_pct}% ({idx + 1}/{total_rows})", end='\r')
            
            person_id = row['person_id']
            index_date = row['INDEX_DATE']
            lookback_limit = lookback_limits.loc[i]
            
            # skip processing if person has no measurements
            if person_id not in grouped_measurements:
                measurement_results.append((np.nan, pd.NaT, np.nan))
                continue
                
            # get all measurements for this person
            person_meas = grouped_measurements[person_id]
            
            # filter by date using vectorized operations
            valid_meas = person_meas[
                (person_meas['measurement_datetime'] <= index_date) &
                (person_meas['measurement_datetime'] >= lookback_limit)
            ]
            
            if valid_meas.empty:
                measurement_results.append((np.nan, pd.NaT, np.nan))
                continue
                
            # calculate days difference using vectorized operations
            valid_meas = valid_meas.copy()
            valid_meas['days_diff'] = (index_date - valid_meas['measurement_datetime']).dt.days
            
            # find closest measurement
            closest_idx = valid_meas['days_diff'].idxmin()
            closest_meas = valid_meas.loc[closest_idx]
            
            measurement_results.append((
                closest_meas['value_as_number'],
                closest_meas['measurement_datetime'],
                closest_meas['days_diff']
            ))
        
        print("\n  Updating result dataframe...")
        # nulk update the result dataframe
        result_df[[f'{measurement_type}_value', f'{measurement_type}_date', f'{measurement_type}_days_diff']] = measurement_results
        
        # show completion for this measurement type
        print(f"  Completed processing {measurement_type} measurements")
    
    print("\nAll measurements processed successfully!")
    return result_df

# using the optimized function
measurement_dfs = {
    'bmi': dat_bmi,
    'tc': dat_tc,
    'hdlc': dat_hdlc
}

final_result = merge_all_measurements(dat_index, measurement_dfs)

# drop helper columns
columns_to_drop = ['bmi_date', 'bmi_days_diff', 'tc_date', 'tc_days_diff', 'hdlc_date', 'hdlc_days_diff']
final_result = final_result.drop(columns=columns_to_drop)

# drop na
final_result = final_result.dropna()

# 14119 patients
print(f"Final dataframe shape: {final_result.shape}")
final_result.head()
```

### Merge Smoker Data

```python
# make copy
final_result = final_result.copy()

# convert to datetime
dataset_76138118_observation_df['observation_datetime'] = pd.to_datetime(dataset_76138118_observation_df['observation_datetime']).dt.tz_localize(None)
final_result.loc[:, 'INDEX_DATE'] = pd.to_datetime(final_result['INDEX_DATE']).dt.tz_localize(None)

# map person id to all smoker record dates
smoker_dict = {}
for person_id, group in dataset_76138118_observation_df.groupby('person_id'):
    smoker_dict[person_id] = group['observation_datetime'].tolist()

# helper function
def has_smoking_record(row):
    person_id = row['person_id']
    index_date = row['INDEX_DATE']
    
    # if not in dict, then 0
    if person_id not in smoker_dict:
        return 0
    
    # get all dates
    obs_dates = smoker_dict[person_id]
    
    # check if any date is within 3 years (1095 days)
    for date in obs_dates:
        if abs((date - index_date).days) <= 1095:
            return 1
    
    return 0

# apply function
final_result.loc[:, 'smoker'] = final_result.apply(has_smoking_record, axis=1)
```

### Merge Antihypertensive Medication Data

```python
# Make a copy of final_result to avoid modifying the original data
final_result = final_result.copy()

# Convert date columns to datetime format and remove timezone information
dataset_76138118_drug_df.loc[:, 'drug_exposure_start_datetime'] = pd.to_datetime(
    dataset_76138118_drug_df['drug_exposure_start_datetime']
).dt.tz_localize(None)

dataset_76138118_drug_df.loc[:, 'drug_exposure_end_datetime'] = pd.to_datetime(
    dataset_76138118_drug_df['drug_exposure_end_datetime']
).dt.tz_localize(None)

final_result.loc[:, 'INDEX_DATE'] = pd.to_datetime(final_result['INDEX_DATE']).dt.tz_localize(None)

# Create a dictionary mapping person_id to all drug exposure start and end dates
drug_exposure_dict = {}
for person_id, group in dataset_76138118_drug_df.groupby('person_id'):
    drug_exposure_dict[person_id] = list(zip(group['drug_exposure_start_datetime'], group['drug_exposure_end_datetime']))

# Helper function to check if a person has drug exposure within the 3-year lookback window
def has_antihypertensive_record(row):
    person_id = row['person_id']
    index_date = row['INDEX_DATE']
    
    # If person_id not in the dictionary, return 0
    if person_id not in drug_exposure_dict:
        return 0

    # Get all drug exposure periods
    exposure_periods = drug_exposure_dict[person_id]
    
    # Check if any exposure period falls within the 3-year (1095 days) lookback window
    for start_date, end_date in exposure_periods:
        if (index_date - pd.Timedelta(days=1095) <= start_date <= index_date) or \
           (index_date - pd.Timedelta(days=1095) <= end_date <= index_date):
            return 1

    return 0

# Apply function to create the 'antihtn' column
final_result.loc[:, 'antihtn'] = final_result.apply(has_antihypertensive_record, axis=1)

# Display the updated DataFrame
final_result.head()
```

```python
# convert to datetime
dat_db['condition_start_datetime'] = pd.to_datetime(dat_db['condition_start_datetime']).dt.tz_localize(None)
final_result.loc[:, 'INDEX_DATE'] = pd.to_datetime(final_result['INDEX_DATE']).dt.tz_localize(None)

# map person id to all db record dates
db_dict = {}
for person_id, group in dat_db.groupby('person_id'):
    db_dict[person_id] = group['condition_start_datetime'].tolist()

# helper function
def has_db_record(row):
    person_id = row['person_id']
    index_date = row['INDEX_DATE']
    
    # if not in dict, then 0
    if person_id not in db_dict:
        return 0
    
    # get all dates
    obs_dates = db_dict[person_id]
    
    # check if any date is within 3 years (1095 days)
    for date in obs_dates:
        if abs((date - index_date).days) <= 1095:
            return 1
    
    return 0

# apply function
final_result.loc[:, 'diabetes'] = final_result.apply(has_db_record, axis=1)
final_result.head()
```

### Merge Events Data

```python
# filter by ICD codes
dat_mi = dataset_76138118_condition_df[
    dataset_76138118_condition_df['source_concept_code'].str.startswith(('410', 'I21', 'I22', 'I23'))]

dat_stroke = dataset_76138118_condition_df[
    dataset_76138118_condition_df['source_concept_code'].str.startswith(('431', '432', '433', '434', 'I61', 'I62', 'I63'))]

dat_hf = dataset_76138118_condition_df[
    dataset_76138118_condition_df['source_concept_code'].str.startswith(('428', 'I50'))]
```

#### Myocardial Infarction

```python
# convert datetime
dat_mi.loc[:, 'condition_start_datetime'] = pd.to_datetime(dat_mi['condition_start_datetime']).dt.tz_localize(None)
final_result.loc[:, 'INDEX_DATE'] = pd.to_datetime(final_result['INDEX_DATE']).dt.tz_localize(None)

# mapping dict
mi_dict = {}
for person_id, group in dat_mi.groupby('person_id'):
    mi_dict[person_id] = group['condition_start_datetime'].tolist()

# helper function
def assign_mi(row):
    person_id = row['person_id']
    index_date = row['INDEX_DATE']
    
    # if person not in mi_dict, assign 0
    if person_id not in mi_dict:
        return 0
    
    # get all dates
    condition_dates = mi_dict[person_id]
    
    # if event before INDEX_DATE, mark for removal
    if any(date < index_date for date in condition_dates):
        return 'REMOVE'
    
    # otherwise, assign 1
    return 1

# apply function to create mi column
final_result.loc[:, 'mi'] = final_result.apply(assign_mi, axis=1)

# remove rows where 'mi' is 'REMOVE'
final_result = final_result[final_result['mi'] != 'REMOVE']

# convert column back to integers
final_result.loc[:, 'mi'] = final_result['mi'].astype(int)
```

```python
# convert datetime
dat_stroke.loc[:, 'condition_start_datetime'] = pd.to_datetime(dat_stroke['condition_start_datetime']).dt.tz_localize(None)
final_result.loc[:, 'INDEX_DATE'] = pd.to_datetime(final_result['INDEX_DATE']).dt.tz_localize(None)

# mapping dict
stroke_dict = {}
for person_id, group in dat_stroke.groupby('person_id'):
    stroke_dict[person_id] = group['condition_start_datetime'].tolist()

# helper function
def assign_stroke(row):
    person_id = row['person_id']
    index_date = row['INDEX_DATE']
    
    # if person not in stroke_dict, assign 0
    if person_id not in stroke_dict:
        return 0
    
    # get all dates
    condition_dates = stroke_dict[person_id]
    
    # if event before INDEX_DATE, mark for removal
    if any(date < index_date for date in condition_dates):
        return 'REMOVE'
    
    # otherwise, assign 1
    return 1

# apply function to create stroke column
final_result.loc[:, 'stroke'] = final_result.apply(assign_stroke, axis=1)

# remove rows where 'stroke' is 'REMOVE'
final_result = final_result[final_result['stroke'] != 'REMOVE']

# convert column back to integers
final_result.loc[:, 'stroke'] = final_result['stroke'].astype(int)
```

#### CVD and ASCVD
