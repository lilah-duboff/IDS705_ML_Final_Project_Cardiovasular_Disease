# Process Data For Logistic

```python
bangladesh_data.drop(
    columns=[
        "Height (cm)",
        "Waist-to-Height Ratio",
        "Weight (kg)",
        "Height (m)",
        "Abdominal Circumference (cm)",
    ],
    inplace=True,
)
```

### Recoding guide

- Sex: 1 for female, 0 for male
- Smoking Status: 1 for yes, 0 for no
- Diabetes Status: 1 for yes, 0 for no
- Physical Activity Level: 0 for low, 1 for moderate, 2 for high
- CVD Risk Level: 1 for intermediary or high risk, 0 for low to no risk
- Blood Pressure Category: 0 for normal; 1 for elevated; 2 for hypertension stage one; and 3 for hypertension stage two
- Family history of CVD: 0 for no; 1 for yes

```python
# Recoding Sex
bangladesh_data["Sex"] = bangladesh_data["Sex"].replace({"F": 1, "M": 0})

# Recoding Smoking Status
bangladesh_data["Smoking Status"] = bangladesh_data["Smoking Status"].replace(
    {"Y": 1, "N": 0}
)

# Recoding Diabetes Status
bangladesh_data["Diabetes Status"] = bangladesh_data["Diabetes Status"].replace(
    {"Y": 1, "N": 0}
)

# Recoding Physical Activity Level
bangladesh_data["Physical Activity Level"] = bangladesh_data[
    "Physical Activity Level"
].replace({"Low": 0, "Moderate": 1, "High": 2})

# Recoding CVD Risk Level
bangladesh_data["CVD Risk Level"] = bangladesh_data["CVD Risk Level"].replace(
    {"LOW": 0, "INTERMEDIARY": 1, "HIGH": 1}
)

# Recoding Blood Pressure Category
bangladesh_data["Blood Pressure Category"] = bangladesh_data[
    "Blood Pressure Category"
].replace(
    {"Normal": 0, "Elevated": 1, "Hypertension Stage 1": 2, "Hypertension Stage 2": 3}
)

# Recoding Family History of CVD
bangladesh_data["Family History of CVD"] = bangladesh_data[
    "Family History of CVD"
].replace({"Y": 1, "N": 0})
```

### Missing Data

Columns that contain missing data include:
- Age
- BMI
- Total Cholesterol
- HDL
- Fasting Blood Sugar
- Systolic BP
- Diastolic BP
- Estimated LDL
- CVD Risk Score

For most of these (excluding CVD risk score), we can impute the missing data by filling in with the mean value of the column. 

```python
# Impute missing values
bangladesh_data["Age"].fillna(bangladesh_data["Age"].mean(), inplace=True)

bangladesh_data["BMI"].fillna(bangladesh_data["BMI"].mean(), inplace=True)

bangladesh_data["Total Cholesterol (mg/dL)"].fillna(
    bangladesh_data["Total Cholesterol (mg/dL)"].mean(), inplace=True
)

bangladesh_data["HDL (mg/dL)"].fillna(
    bangladesh_data["HDL (mg/dL)"].mean(), inplace=True
)

bangladesh_data["Fasting Blood Sugar (mg/dL)"].fillna(
    bangladesh_data["Fasting Blood Sugar (mg/dL)"].mean(), inplace=True
)

bangladesh_data["Systolic BP"].fillna(
    bangladesh_data["Systolic BP"].mean(), inplace=True
)

bangladesh_data["Diastolic BP"].fillna(
    bangladesh_data["Diastolic BP"].mean(), inplace=True
)

bangladesh_data["Estimated LDL (mg/dL)"].fillna(
    bangladesh_data["Estimated LDL (mg/dL)"].mean(), inplace=True
)

bangladesh_data.isnull().sum()
```

```python
bangladesh_data = bangladesh_data.dropna(subset=["CVD Risk Score"])

bangladesh_data.isnull().sum()
```
